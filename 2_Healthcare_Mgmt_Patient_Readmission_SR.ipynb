```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as sp
```

```python
# Load the dataset
df = pd.read_csv("diabetic_data.csv")
```

3.1) Perform descriptive statistics analysis for numerical features

```python
numeric_stats = df.select_dtypes([float,int]).describe()
print("Descriptive Statistics for Numerical Features:")
numeric_stats
```

3.2) Visualize the distribution of categorical features - race and gender


```python
categorical_features = ['race', 'gender']
for feature in categorical_features:
    plt.figure(figsize=(6, 4))
    df[feature].value_counts().plot(kind='bar', color='orange')
    plt.title(f'Distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Count')
    plt.show()
```

3.3) Explore relationship between readmission status and age

```python
counts = pd.crosstab(df.age,df.readmitted)
plt.figure(figsize=(10, 6))
counts.plot(kind='bar')
plt.title('Relationship between Age and Readmission Status')
plt.show()
```

3.4)Investigate correlations between numerical features

```python
correlation_matrix = df.select_dtypes([float,int]).corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", annot_kws={"size": 10})
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

3.5) Analyze distribution of medication changes and total medications taken

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df,x='change', y='num_medications')
plt.title('Distribution of Medication Changes and Total Medications Taken')
plt.xlabel('Count')
plt.ylabel('Frequency')
plt.legend(['Medication Changes', 'Total Medications Taken'])
plt.show()
```

3.6) Examine distribution of diagnoses categories

```python
diagnosis_features = ['diag_1', 'diag_2', 'diag_3']
for feature in diagnosis_features:
    plt.figure(figsize=(6, 4))
    df[feature].value_counts().plot(kind='bar', color='green')
    plt.title(f'Distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Count')
    plt.show()
```

3.7) Explore distribution of patients across admission types, sources, and discharge dispositions

```python
admission_features = ['admission_type_id', 'admission_source_id', 'discharge_disposition_id']
for feature in admission_features:
    plt.figure(figsize=(8, 6))
    df[feature].value_counts().plot(kind='bar', color='skyblue')
    plt.title(f'Distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Count')
    plt.show()
```

3.8) Identify and visualize any outliers in the dataset, especially in numerical features

```python
sns.boxplot(x='admission_type_id', y='time_in_hospital', data=df)
plt.show()
```
